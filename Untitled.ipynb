```python
import numpy as np 
import sys
sys.path.append("test")
import defs
from PIL import Image
from sealwatch.cost_estimator._attack import attack
cover = np.array(Image.open(defs.COVER_UNCOMPRESSED_GRAY_DIR / "seal7.png"))
for alpha in [0.7, 0.8]:
    for cf in defs.COST_FUNCTIONS:
        stego = cf[1](cover, alpha)
        res = attack(stego, cover)
        print(alpha, "true:", cf[0], "-> picked:", res["method"])
        for r in sorted(res["all"], key=lambda x: -x["log_lik"]):
            print(f"   {r['method_name']:9s} log_lik={r['log_lik']:.2f}")
```
